```python
import joblib
import pandas as pd
from google.colab import drive
```

```python
# 1. Mount Google Drive
drive.mount('/content/drive')
```

```python
# Define the path to your saved model
model_path = '/content/drive/MyDrive/rvs_genai_course/genai_projects/Health_Insurance_Prediction_ML/linear_regression_model.joblib'
```

```python
# 2. Load the model
try:
    loaded_model = joblib.load(model_path)
    print("Model loaded successfully!")
except FileNotFoundError:
    print(f"Error: Model file not found at {model_path}. Please check the path and ensure Drive is mounted.")
    # Exit or handle the error appropriately
    exit()
```

```python
# 3. Prepare new input data for prediction
# This DataFrame must have the same columns and order as X_train during training.
# Based on X.columns in the notebook: ['age', 'bmi', 'children', 'sex_male', 'smoker_yes', 'region_northwest', 'region_southeast', 'region_southwest']

# Example new data for a male, 30 years old, BMI 25, 1 child, non-smoker, in the northeast region
# For categorical features, remember the one-hot encoding logic used (e.g., sex_male = 1 for male, 0 for female)
# And 'northeast' was the baseline region, so its dummy variables would all be 0.
```

```python
#For Getting Input from user

# age = int(input("Enter the age: "))
# sex = input("Enter the sex (male/female): ")
# bmi = float(input("Enter the BMI: "))
# children = int(input("Enter the number of children: "))
# smoker = input("Enter the smoking status (yes/no): ")
# region = input("Enter the region (southeast/southwest/northeast/northwest): ")

def get_input_df(df_new, age, sex, bmi, children, smoker, region):

    df_new.loc[0, 'age'] = age
    df_new.loc[0, 'sex_male'] = 1 if sex.lower() == 'male' else 0
    df_new.loc[0, 'bmi'] = bmi
    df_new.loc[0, 'children'] = children
    df_new.loc[0, 'smoker_yes'] = 1 if smoker.lower() == 'yes' else 0
    df_new.loc[0, 'region_northwest'] = 1 if region.lower() == 'northwest' else 0
    df_new.loc[0, 'region_southwest'] = 1 if region.lower() == 'southwest' else 0
    df_new.loc[0, 'region_southeast'] = 1 if region.lower() == 'southeast' else 0


df_new = pd.DataFrame(columns=['age', 'bmi', 'children', 'sex_male', 'smoker_yes', 'region_northwest', 'region_southeast', 'region_southwest'])

get_input_df(df_new, 54, 'female', 31.9, 3, 'no', 'southwest')

#prediction

data_to_pred = df_new.head(1)

res = loaded_model.predict(data_to_pred)

print("Prepared input DataFrame for prediction:")
display(df_new.head())

print(f"\nPredicted charges for the input: {res[0]:.2f}")
```
